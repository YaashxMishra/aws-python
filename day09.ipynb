# change values

```python
# normal array
l=[10,20,30,40]
l[0]=100
l[-1]=150
print(l)

# numpy array
import numpy as np
arr=np.array(l)
arr[0]=200
arr[-1]=250
print(arr)
```

# Arithmetic Operations on NumPy Arrays.

```python
l=[10,20,30,40]
# print(l+2) TypeError: can only concatenate list (not "int") to list
# print(l-2) TypeError: can only concatenate list (not "int") to list
# print(l/2) TypeError: can only concatenate list (not "int") to list
print(l*2) # only this works

import numpy as np
arr=np.array(l)
print(arr+2)
print(arr-2)
print(arr*2)
print(arr/2)
```

# 2D array using arange

```python
import numpy as np
arr=np.arange(6).reshape(2,3) # 2 rows and 3 columns
# rows * columns = total number of elements. Here 2 * 3 = 6
print(arr)
```

```python
import numpy as np
arr=np.arange(12).reshape(3,4) # 3 rows and 4 columns
# rows * columns = total number of elements. Here 3 * 4 = 12

print(arr)
```

```python
# custom start,stop and step
arr=np.arange(10,22,2).reshape(3,2)
print(arr)
```

```python
# Automatic column calculation
arr=np.arange(12).reshape(3,-1)
print(arr)
```

```python
# Automatic row calculation
arr=np.arange(12).reshape(-1,4)
print(arr)
```
