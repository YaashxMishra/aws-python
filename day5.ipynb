# **List Methods**

```python
# 1. append()

lis=[1,2,3]
lis.append(4)
print(lis)
```

```python
# 2. insert()
lis=[1,2,3]
lis.insert(1,4)
print(lis)
```

```python
# 3. len()
len(lis)
```

# **Dictionary Methods**

```python
# 1. clear()
d={'name':"yash","age":20}
d.clear()
print(d)
```

```python
# 2. copy()
d={'name':"yash","age":20}
dic=d.copy()
print(dic)
```

```python
# 3. fromkeys()
dic={}
print(dic.fromkeys([1,2,3]))
```

```python
print(dic.fromkeys([1,2,3],0))
```

```python
print(dic.fromkeys([1,2,3],[1,2,3]))
```

```python
# 4. del d[key]
d={'name':"yash","age":20}
del d['age']
print(d)
```

# **For loop**

```python
for i in range(5):
  print(i)
```

```python
lis=[10,20,30,40]
print("index   value")
for i in range(len(lis)):
  print(" ",i,"\t",lis[i])
```

```python
d={'Users':["Yash","Dhruv"]}
for i in d['Users']:
  print(i)
```

# **Functions**

```python
def hello():
  print("Hello function")
hello()
```

```python
def name(n):
  print(n)
name("Yash")
```

```python
def name(a,b):
  print(a,b)
name("yash","dhruv")
```

```python
def name(*a):
  print(a)
name("yash")
name("yash",'dhruv')
name('a','b','c','d','e')
```

```python
def name(*a):
  for i in a:
    print(i)
name("yash")
name("yash",'dhruv')
name('a','b','c','d','e')
```
