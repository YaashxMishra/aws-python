# **Input from the user**

```python
roll_no=input("Enter your roll number : ")
print(roll_no)
print(id(roll_no))
print(type(roll_no))
```

# **Type conversion**

```python
roll_no=int(input("Enter your roll number : ")) # Explicit type conversion
print(roll_no)
print(id(roll_no))
print(type(roll_no))
```

```python
roll_no=float(input("Enter your roll number : "))
print(roll_no)
print(id(roll_no))
print(type(roll_no))
```

```python
roll_no=bool(input("Enter your roll number : "))
print(roll_no)
print(id(roll_no))
print(type(roll_no))
```
