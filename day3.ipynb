# **Operators**
1. Arithmetic -> +,-,*,/, // , % ,**

2. Relational -> >, <, >= , <= , == , !=

3. Assignment -> +=, -= , *= , /=, %=

4. Logical -> and , or , not

5. Bitwise -> &, |

# **Arithmetic Operator**

```python
# Addition
num1=10
num2=20

print(num1 + num2)
```

```python
# Subtraction
num1=100
num2=40

print(num1-num2)
```

```python
# Multiplication
num1=5
num2=2
print(num1 * num2)
```

```python
# Division/Divide
num1=100
num=3

print(100/3)
```

```python
# Modulus
num1=34
num2=5
print(num1 % num2)
```

```python
# Integer division / floar division
num1=34
num2=5

print(num1//num2)
```

```python
# Power operator
num1=2
num2=5

print(num1 ** num2)
```

# **Relational Operator**

```python
# > / Greater than operator
num1=10
num2=20
print(num1 > num2)
```

```python
# < / Less than operator
num1=10
num2=20
print(num1 < num2)
```

```python
# >= / Greater than or equal to operator
num1=10
num2=20
print(num1 >= num2)
```

```python
# <= / Less than or equal to operator
num1=10
num2=20
print(num1 <= num2)
```

```python
# == / equal to operator
num1=10
num2=20
print(num1 == num2)
```

```python
# != / not equal to operator
num1=10
num2=20
print(num1 != num2)
```

# **Assignment Operator**

```python
# += operator
num1=10
num2=20
num1+=num2
print(num1)
```

```python
# -= operator
num1=10
num2=20
num1-=num2
print(num1)
```

```python
# *= operator
num1=10
num2=20
num1*=num2
print(num1)
```

```python
# /= operator
num1=10
num2=20
num1/=num2
print(num1)
```

```python
# %= operator
num1=10
num2=20
num1%=num2
print(num1)
```
