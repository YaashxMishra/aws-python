# **Inheritance**

```python
class user1:
  def __init__(self,pen,pencil) -> None:
    print("Constructor of user1")
    self.pen=pen
    self.pencil=pencil

  def penGet(self):
    print("pen is working :",self.pen)

  def pencilGet(self):
    print("pencil is working :",self.pencil)

class user2(user1):
  def __init__(self,pen,pencil,eraser) -> None:
    print("Constructor of user2")
    super().__init__(pen,pencil)
    self.eraser=eraser


u=user2('blue','black','doms')
u.penGet()

```

```python
class Math:
  def __init__(self) -> None:
    pass

  def pow(self,a,b):
    return a**b

  def factorial(self,n):
    count=1
    if n<0:
      return "factorial of negative number is not defined"
    else:
      for i in range(2,n+1):
        count*=i

    return count

m=Math()
print(m.pow(4,2))
print(m.factorial(5))
print(m.factorial(-1))


```

```python
class Yash:
  def __init__(self,bike):
    self.bike=bike

  def useBike(self):
    print("Bike is :",self.bike)

class Rajendra(Yash):
  def __init__(self,name, bike):
    super().__init__(bike)
    self.name=name

  def useCar(self):
    print("car is used by :",self.name)


u=Rajendra('yash','Yamaha MT-15')
u.useBike()
u.useCar()
```

# **Encapsulation**

```python
class Student:
  def __init__(self,name,roll_no) -> None:
    self.__data=name
    self.roll_no=roll_no
  def get_name(self,password=None):
    if password=='123456':
      print(self.__data)
    else:
      print("password not match")

s=Student('Yash',123)
# print(s.__data) __data is private and Python changes the name internally to: _Student__name and it's called name mangling
print(s._Student__data) # But it can be accessed through this
print(s.roll_no)
s.get_name('123')
s.get_name('123456')
```

# **Class Variable**

```python
class Student:
  count=0 # count is class variable
  def __init__(self,name,roll_no) -> None:
    print("Constructor is working")
    self.name=name
    self.roll_no=roll_no
    # count=13 local variabel or instance variable
    Student.count=12 # you can update the value of count by using class name
    # print(count) will give error
    print(Student.count)
    print(self.count)

  def get_name(self):
    print(self.name)


s=Student('Yash',123)
s.get_name()
print(s.count)
```

# **Overloading**

```python
def Student123(name, roll_no):  # First definition (will be overwritten)
    print(name, roll_no)

def Student123(name, roll_no, age):  # Replaces the previous function
    print(name, roll_no, age)

Student123("Yash", 123, 20)  # Calls the latest definition with 3 arguments
# Student123("Yash", 123)    # TypeError: age argument is missing
```

```python
def Student123(name, roll_no, age=None):
    if age is None:
        print(name, roll_no)
    else:
        print(name, roll_no, age)

Student123("Yash", 123)
Student123("Yash", 123, 20)
```

```python
def Student123(*args):
    print(*args)

Student123("Yash", 123)
Student123("Yash", 123, 20)
```

# **Overriding**

```python
class User1:
  def __init__(self,pen,pencil):
    self.pen=pen
    self.pencil=pencil

  def get_pen(self):
    print("User1:",self.pen)

class User2(User1):
  def __init__(self,pen,pencil,eraser) -> None:
    self.eraser=eraser
    super().__init__(pen,pencil)

  def get_pen(self):
    super().get_pen()
    print("User2:",self.pen)


u=User2('blue','black','doms')
u.get_pen()
print(u.pen)
print(u.pencil)
print(u.eraser)
```
