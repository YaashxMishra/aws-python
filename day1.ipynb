```python
# AND operator returns the first falsy value it encounters.
# If all values are truthy, it returns the last value.

# Example 1
num1 = 10      # Truthy
num2 = 20      # Truthy

print(num1 and num2)  # Output: 20 (both are truthy, so returns the last value)


# Example 2
num1 = []      # Falsy (empty list)
num2 = 20      # Truthy

print(num1 and num2)  # Output: [] (returns the first falsy value)


# Example 3
num1 = [1]     # Truthy (non-empty list)
num2 = ""      # Falsy (empty string)

print(num1 and num2)  # Output: "" (returns the first falsy value encountered)
```
