# ** Conditional Statements**

```python
# if
num1=10
if num1>2:
  print("num1 = ",num1)
print("if end")
```

```python
# if else
num1=10
num2=20

if num1<=num2:
  print("If part runs")
else:
  print("Else part run")

```

```python
# if elif else
num1=10
num2=20
num3=30

if num1>num2:
  print("If part run")
elif num1 >num3:
  print("elif part run")
else:
  print("else part run")
```

```python
marks=int(input("Enter your marks : "))

if marks>95:
  print("A grade")
elif marks>90:
  print("B grade")
elif marks>70:
  print("C grade")
elif marks>60:
  print("D grade")
else:
  print("Fail")
```

# **List**

```python
l=[1,2,3,4,5]
print(l[0])
print(l[1])
print(l[2])
print(l[3])

# negative indexing
print(l[-1])
print(l[-2])
print(l[-3])

#mutuable
l[0]=100
l[-1]=200
print(l)
```

# **Dictionary**

```python
dic={"name":"yash","age":19}
print(dic['name'])
print(dic["age"])

for key in dic.keys():
  print(key)

for value in dic.values():
  print(value)

print(dic.keys())
print(dic.values())
print(dic.items())

dic['name']="Dhruv"
print(dic)
```
